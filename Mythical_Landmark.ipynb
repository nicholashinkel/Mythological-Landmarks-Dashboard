**Mythical Landmarks**

#### This project uses a dataset of figures from the *Odyssey*, the locations connected to them, coordinates for those locations, and citations for the figure-location connections.

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
import folium as fm
import plotly.express as px

# read the data file and create a data frame
df = pd.read_excel('Mythical Landmark Dataset1.xlsx')
```

#### Map of Locations

```python
# define the map
from google.colab import userdata

key = userdata.get("key")

full_map = fm.Map(
    location=[df['Lat'].mean(), df['Lon'].mean()],
    zoom_start=5,
    tiles=f"https://basemaps.cartocdn.com/rastertiles/dark_all/{{z}}/{{x}}/{{y}}.png?key={key}",
    attr="© OpenStreetMap contributors © CARTO"
)
# group figures into groups of same location
for loc, group in df.groupby('Location'):
    grouped_figures_by_loc = ', '.join(group['Figure'].unique())

    # create map
    fm.CircleMarker(
        location = [group['Lat'].iloc[0], group['Lon'].iloc[0]],
        radius = 5,
        fill = True,
        color = '#2288b3',
        popup = grouped_figures_by_loc,
        tooltip = loc
    ).add_to(full_map)

# display the map
full_map
```

#### Locations by Type

```python
from google.colab import userdata

key = userdata.get("key")

# create a map colored by location type
loc_type_map = fm.Map(
    location=[df['Lat'].mean(), df['Lon'].mean()],
    zoom_start=5,
    tiles=f"https://basemaps.cartocdn.com/rastertiles/dark_all/{{z}}/{{x}}/{{y}}.png?key={key}",
    attr="© OpenStreetMap contributors © CARTO"
)

for i, row in df.iterrows():
    if row['Loc_type'] == 'Physical':
        marker_color = '#5D3FD3'
    else:
        marker_color = '#ff3355'

    fm.CircleMarker(
        location=[row['Lat'], row['Lon']],
        radius=5,
        color=marker_color,
        tooltip=row['Location'] + ' : ' + row['Loc_type']
    ).add_to(loc_type_map)

# display the map
loc_type_map
```

#### Most Common Locations

```python
# count how many figures are connected to each location
loc_count = df.groupby('Location')['Figure'].nunique().reset_index()
loc_count.rename(columns = {'Figure': 'fig_count'}, inplace = True)

# sort the values
loc_count = loc_count.sort_values('fig_count', ascending = False)
loc_count.head()
```

```python
# create bar chart of most common locations
loc_fig_chart = loc_count.head(30)

fig_loc_fig_chart = px.bar(
    loc_fig_chart,
    x = 'fig_count',
    y = 'Location',
    orientation = 'h',
    title = 'Locations with Most Figures',
    text = 'fig_count'
)

fig_loc_fig_chart.update_layout(
    xaxis_title = 'Amount of Figures',
    yaxis_title = 'Location')

fig_loc_fig_chart.update_traces(marker_color = "#2288b3")
fig_loc_fig_chart.update_layout(title_x = 0.5)
fig_loc_fig_chart.update_layout(template = "plotly_dark")
fig_loc_fig_chart.update_yaxes(autorange = 'reversed')

fig_loc_fig_chart.show()
```

#### Figure and Location Graph

```python
# create the graph
G = nx.Graph()

for i, row in df.iterrows():
    G.add_edge(row['Figure'], row['Location'])

# print the number of vertices and edges
print(G)
```

```python
# draw the graph
plt.figure(figsize = (15, 10))

pos = nx.spring_layout(G, seed = 42)

nx.draw(
    G,
    pos,
    font_size = 9,
    node_size = 2500,
    node_color = '#2288b3',
    with_labels = True
)

plt.title('Figures to Locations')
plt.show()
```

#### Network Graph

```python
# install PyVis
!pip install pyvis

from pyvis.network import Network
```

```python
# create network graph
net = Network(
    notebook = True,
    height= 700,
    width= '100%',
    bgcolor='black',
    font_color='white',
    cdn_resources='in_line'
)

net.from_nx(G)

for node in net.nodes:
    node["color"] = "#2288b3"
    node_name = node["label"]
    degree = G.degree[node_name]
    node["size"] = 10 + degree * 5

# show the graph
net.show('network_graph.html')
```

#### Odysseus Journey

```python
# create a dataframe of Odysseus' locations
odysseus_df = df[df['Figure'] == 'Odysseus'].copy()

# sort by in order of 'Stop'
odysseus_df = odysseus_df.sort_values('Stop')

# simple sanity check
odysseus_df.head()
```

```python
from google.colab import userdata

key = userdata.get("key")

# create a map of the journey
fig_line = px.line_mapbox(
    odysseus_df,
    lat='Lat',
    lon='Lon',
    hover_name='Location',
    text='Location',
    zoom=4.5,
    height=650,
    hover_data=["Figure", "Description"],
    color_discrete_sequence=["#2288b3"]
)

fig_line.update_layout(
    mapbox_style={
        "version": 8,
        "sources": {
            "carto": {
                "type": "raster",
                "tiles": [
                    f"https://basemaps.cartocdn.com/rastertiles/dark_all/{{z}}/{{x}}/{{y}}.png?key={key}"
                ],
                "tileSize": 256,
                "attribution": "© OpenStreetMap contributors © CARTO"
            }
        },
        "layers": [
            {"id": "carto", "type": "raster", "source": "carto"}
        ]
    }
)

fig_line.show()
```

#### Animated Journey

```python
from google.colab import userdata
import plotly.express as px

key = userdata.get("key")

# Animated map of Odysseus' Journey
fig_anim = px.scatter_mapbox(
    odysseus_df,
    lat='Lat',
    lon='Lon',
    hover_name='Location',
    text='Location',
    animation_frame='Stop',
    zoom=3,
    height=650,
    color_discrete_sequence=["#2288b3"]
)

# Set background using CARTO tiles
fig_anim.update_layout(
    mapbox_style="white-bg",
    mapbox_layers=[
        {
            "sourcetype": "raster",
            "source": [
                f"https://basemaps.cartocdn.com/rastertiles/dark_all/{{z}}/{{x}}/{{y}}.png?key={key}"
            ],
            "sourceattribution": "© OpenStreetMap contributors © CARTO",
            "below": "traces"
        }
    ]
)

# Animation speed
fig_anim.layout.updatemenus[0].buttons[0].args[1]["frame"]["duration"] = 1500
fig_anim.layout.updatemenus[0].buttons[0].args[1]["transition"]["duration"] = 700

fig_anim.update_traces(
    textfont=dict(color="white")
)

fig_anim.layout.updatemenus[0].font = dict(color="black")

fig_anim.show()
```

### What is an API Key?
An API (Application Programming Interface) key is a unique identifier used to authenticate a user, developer, or calling program to an API. It's like a password that grants you access to use a particular service.

### Why is it required?
Many services require API keys to:

1.  **Identify and authenticate** users.
2.  **Control access** to their services, often to prevent misuse or manage usage limits.
3.  **Track usage** for billing or analytics purposes.

### How to get and use an API Key in Colab:
If you're using a service like Google's Gemini API, you typically need to:

1.  **Obtain the API Key**: Go to the service provider's developer console (e.g., Google AI Studio for Gemini) and generate a new API key.
2.  **Store it securely in Colab**: In Google Colab, it's best practice to store sensitive information like API keys in the "Secrets" manager. You can access it by clicking the "🔑" icon in the left sidebar. Give your secret a name (e.g., `GOOGLE_API_KEY`).
3.  **Access the key in your code**: You can then retrieve this secret in your Python code using `from google.colab import userdata` and then `api_key = userdata.get('YOUR_SECRET_NAME')`.
